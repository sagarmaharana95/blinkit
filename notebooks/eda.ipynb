```python
tables = ["customers", "products", "orders", "order_items", 
          "delivery_performance", "customer_feedback", "inventory", "marketing_performance"]

dataframes = {}
for t in tables:
    dataframes[t] = pd.read_sql(f"SELECT * FROM {t}", engine)
    print(f"{t}: {dataframes[t].shape}")
```

```python
customers = dataframes['customers']
products = dataframes['products']
orders = dataframes['orders']
order_items = dataframes['order_items']
delivery_performance = dataframes['delivery_performance']
customer_feedback = dataframes['customer_feedback']
inventory = dataframes['inventory']
marketing_performance = dataframes['marketing_performance']
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(orders['order_total'], bins=40, kde=True, ax=axes[0], color='steelblue')
axes[0].set_title('Order Total Distribution')

sns.histplot(delivery_performance['delivery_time_minutes'], bins=30, kde=True, ax=axes[1], color='orange')
axes[1].set_title('Delivery Time Distribution')

sns.histplot(products['price'], bins=30, kde=True, ax=axes[2], color='green')
axes[2].set_title('Product Price Distribution')

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(x=orders['order_total'], ax=axes[0], color='steelblue')
axes[0].set_title('Order Total — Outlier Check')

sns.boxplot(x=delivery_performance['delivery_time_minutes'], ax=axes[1], color='orange')
axes[1].set_title('Delivery Time — Outlier Check')

plt.tight_layout()
plt.show()
```

```python
# Merge relevant numeric columns for correlation
merged = orders.merge(delivery_performance, on='order_id', suffixes=('_order', '_delivery'))

numeric_cols = merged[['order_total', 'delivery_time_minutes', 'distance_km']]

plt.figure(figsize=(6, 5))
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt='.2f', vmin=-1, vmax=1)
plt.title('Correlation Matrix — Order & Delivery Metrics')
plt.show()
```

```python
# Top product categories by revenue
category_revenue = order_items.merge(products, on='product_id') \
    .assign(revenue = lambda x: x['quantity'] * x['unit_price']) \
    .groupby('category')['revenue'].sum() \
    .sort_values(ascending=False)

plt.figure(figsize=(10, 5))
category_revenue.plot(kind='bar', color='teal')
plt.title('Revenue by Product Category')
plt.ylabel('Revenue (₹)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Marketing channel performance - ROAS comparison
channel_performance = marketing_performance.groupby('channel').agg(
    total_spend=('spend', 'sum'),
    total_revenue=('revenue_generated', 'sum'),
    avg_roas=('roas', 'mean')
).sort_values('avg_roas', ascending=False)

print(channel_performance)
```
